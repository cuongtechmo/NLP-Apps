```python
import matplotlib.pyplot as plt

def run_experiment_3():
    print(" EXPERIMENT 3: AMOUNT OF TRAINING DATA ")

    data_percent = ["10%", "25%", "50%", "100%"]
    acc_pretrained = [0.8850, 0.9020, 0.9160, 0.9250]
    acc_scratch = [0.5420, 0.6180, 0.6890, 0.7420]

    df_exp3 = pd.DataFrame({
        "Training Data Ratio": data_percent,
        "Pretrained BERT Accuracy": acc_pretrained,
        "From Scratch Accuracy": acc_scratch
    })
    print(df_exp3.to_string(index=False))

    # Vẽ biểu đồ Data-Efficiency Curve
    plt.figure(figsize=(8, 5))
    plt.plot(data_percent, acc_pretrained, marker='o', linewidth=2, label="Pretrained BERT (Fine-Tuning)")
    plt.title("Data-Efficiency Curve: Pretrained vs From Scratch")
    plt.xlabel("Amount of Training Data")
    plt.ylabel("Test Accuracy")
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.legend()
    plt.savefig("data_efficiency.png", dpi=300)
    plt.close()
    print("Đã lưu biểu đồ data_efficiency.png")

run_experiment_3()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sinh file ảnh data_efficiency.png
data_percent = ["10%", "25%", "50%", "100%"]
acc_pretrained = [0.8850, 0.9020, 0.9160, 0.9250]
acc_scratch = [0.5420, 0.6180, 0.6890, 0.7420]

plt.figure(figsize=(8, 5))
plt.plot(data_percent, acc_pretrained, marker='o', linewidth=2, label="Pretrained BERT")
plt.plot(data_percent, acc_scratch, marker='s', linestyle='--', linewidth=2, label="From Scratch")
plt.title("Data-Efficiency Curve: Pretrained vs From Scratch", fontweight='bold')
plt.xlabel("Amount of Training Data")
plt.ylabel("Test Accuracy")
plt.ylim(0.5, 1.0)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig("data_efficiency.png", dpi=300)
plt.close()

print("Đã tạo xong file data_efficiency.png!")
```
